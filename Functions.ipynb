```python
def sum(a, b):# parameter
    print(a+b)
sum(5,10)# arguement

```

```python
def vote(age):
    if age>18:
        print("eligible to vote")
    else:
        print("eligible not")
vote(20)
```

```python
def average(a,b,c,d):
    print((a+b+c+d)/4)
avg=average(4,5,5,6)
print(avg);
```

```python
def mycountry(country):
    return f'I Love {country} '
print(mycountry("INDIA"))
print(mycountry("Londan"))
```

```python
def square(a):
    return a**a
square(4)
```

```python
square=lambda a:a**a
print(square(4))
```

```python
def number(num):
    if num%2==0:
        print("even")
    else:
        print("odd")
number(12)
```
