# What is a Data Type?  
-   Data type specifies the  type of value a variable holds the value of variable.
-   This is required in programming to do various operations without causing an error. 

#### These are four variables of different data types. 
In python, we can print the type of any operator using type function: 
 












```python
a=12
b=7.8
c=2.5
print(type(a))
print(type(b))
print(type(c))
```

#####   By default,python provides the following built in data types:
#### 1. Numeric data: 
-   ##### int, float, complex 
-   int: 3 ,3 -   float: 7.349, -9.0, 0.0000001 -  complex: 6 + 2i , 
#### 2. Text data: 
-   ##### Str 
-   Str: “Hello World!!!”, “Python Programming” 
#### 3. Boolean data: 
-   Boolean data consists of values True or False. 
#### 4.Sequenced data:
-   ##### list,tuples:
#### List
-   A list is an ordered collection of data with elements separated by a comma and enclosed within square brackets. 
-   Lists are mutable and can be modified after creation.
#### Examples: 

```python
list1 = [8, 2.3, [-4, 5], ["apple", "banana"]]
print(list1)
```

### Tuple:
-    A tuple is an ordered collection of data with elements separated by a comma and enclosed within parentheses. Tuples are immutable and can not be modified after creation.  

### Examples:


```python
tuple1 = (("parrot", "sparrow"), ("Lion", "Tiger")) 
print(tuple1) 
```

### 5. Mapped data: 
####    Dictionary 
-   A dictionary is an unordered collection of data containing a key:value pair. The key:value pairs are enclosed within curly brackets. 


```python
dict1 = {"name":"Sakshi", "age":20, "canVote":True} 
print(dict1) 
```
