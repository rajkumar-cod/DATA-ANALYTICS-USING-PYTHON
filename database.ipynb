```python
from sqlalchemy import create_engine, Column, Integer, String
from sqlalchemy.orm import declarative_base, sessionmaker
```

-ceate_engine:it used to make connection from the declarative_base
-column: it defines columns in declarative_base
-Integer  and Sting: datatype of values in database

```python
engine=create_engine('sqlite:///user.db')
```

```python
Base=declarative_base()
```

```python
class User(Base):
    __tablename__='users'
    id=Column(Integer, primary_key=True)
    name=Column(String)
    password=Column(String)
```

```python
Base.metadata.create_all(engine)
```

```python
Session=sessionmaker(bind=engine)
Session=Session()
```

```python
new_user=User(name='Alice', password='2736') # add user
Session.add(new_user)
Session.commit()
```

```python
users=Session.query(User).all() # retrieve all users

    

# Print user details
for user in users:

    print(f'ID: {user.id}, Name: {user.name}, Password: {user.password}')
```

```python
#update the useruser
user=Session.query(User).filter_by(id=1).first()
user.name='John Smith'
Session.commit()
```
