```python
import seaborn as sns
```

```python
df=sns.load_dataset('titanic')
df
```

```python
# Age Distribution
fig=sns.histplot(df, x='age')
fig.set_title('Age Distribution of titanic Passenger')
fig.set_xlabel("Age")
fig.set_ylabel("frequency")
```

```python
# survival counts based on gender
fig=sns.countplot(df, x='survived')
fig.set_title('Survival Counts based on gender')
fig.set_xlabel("Survived")
fig.set_ylabel("Count")
```

```python
fig=sns.barplot(df, x='sex', y='fare', hue='class')
fig.set_title('Averge Fare by Pssenger Class and Gender ')
fig.set_xlabel('Sex')
fig.set_ylabel('Average Fare')
```

```python
fig=sns.barplot(df, x='sex', y='fare', palette='coolwarm')
fig.set_title('Count of  Pssenger by Class  ')
fig.set_xlabel('Passenger Class')
fig.set_ylabel('Count')
```
